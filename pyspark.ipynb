```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.window import Window
spark = SparkSession.builder.appName("practice").getOrCreate()
data = [
(1, "Alice", "Engineering", 95000, "NY", "2019-03-01"),
(2, "Bob", "Engineering", 88000, "NY", "2020-07-15"),
(3, "Carla", "Engineering", 102000, "CA", "2018-01-10"),
(4, "David", "Sales", 65000, "TX", "2021-06-01"),
(5, "Eva", "Sales", 71000, "TX", "2019-11-20"),
(6, "Frank", "Sales", 71000, "NY", "2022-02-14"),
(7, "Grace", "HR", 58000, "CA", "2020-09-09"),
(8, "Heidi", "HR", 60000, "CA", "2017-05-05"),
(9, "Ivan", "Engineering", 99000, "TX", "2023-01-01"),
(10, "Judy", None, 45000, "NY", "2021-08-08"),
]
columns = ["id", "name", "dept", "salary","state", "hire_date"]
df = spark.createDataFrame(data, columns)
df.show()
```

```python
def reverse_string(s):
    result = ""
    for ch in s:
        result = ch + result
    return result
```

```python
user_input_string = input("Enter a string to reverse: ")
reversed_user_string = reverse_string(user_input_string)
print(f"The original string was: {user_input_string}")
print(f"The reversed string is: {reversed_user_string}")
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructField,StructType,StringType,IntegerType,DataType

spark = SparkSession.builder.appName("practice").getOrCreate()

schema= StructType([
    StructField("id",IntegerType(),True),
    StructField("Name",StringType(),True)
])

# Create a blank DataFrame with the schema
df = spark.createDataFrame([], schema=schema)

# Sample data to add
new_data = [
    (101, "Alice"),
    (102, "Bob")
]

# Create a new DataFrame from the sample data with the same schema
new_data_df = spark.createDataFrame(new_data, schema=schema)

# Add data to the existing DataFrame using unionByName
df = df.unionByName(new_data_df)
df.show()
```

```python
# 1. Handle missing department values by filling with 'Unknown'
df_filled_dept = df.fillna({'dept': 'Unknown'})

# 2. Group by department and calculate the average salary
average_salaries_df = df_filled_dept.groupBy('dept').agg(F.avg('salary').alias('average_salary'))

# Display the average salaries per department
print("Average Salary per Department:")
average_salaries_df.show()

# 3. Identify the department with the highest average salary
highest_avg_salary_dept = average_salaries_df.orderBy(F.desc('average_salary')).first()

print(f"Department with the Highest Average Salary: {highest_avg_salary_dept['dept']} (Average Salary: {highest_avg_salary_dept['average_salary']:.2f})")
```
