```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("shivamb/netflix-shows")

print("Path to dataset files:", path)
```

```python
import os
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName('Netflix').getOrCreate()
print(f"Files in {path}:\n{os.listdir(path)}")
```

```python
# Construct the full path to the CSV file
csv_file_path = os.path.join(path, "netflix_titles.csv")

# Read the CSV file into a PySpark DataFrame
df_spark = spark.read.csv(csv_file_path, header=True, inferSchema=True)

# Display the first 5 rows and the schema
df_spark.show(5)
df_spark.printSchema()
```

### 5 Easy Practice Questions on PySpark Transformations

Based on your Netflix dataset (`df_spark`), try writing code to solve these basic transformation exercises:

1. **Filtering Rows**
   How do you filter the DataFrame to show only rows where the `type` is exactly `'Movie'`?

2. **Selecting Columns**
   How do you create a new DataFrame containing only the `title`, `director`, and `release_year` columns?

3. **Renaming a Column**
   How do you rename the column `release_year` to `year_of_release`?

4. **Dropping a Column**
   How do you drop the `description` column entirely from the DataFrame?

5. **Creating a Calculated Column**
   How do you create a new column named `is_recent` that contains `True` if `release_year` is greater than or equal to `2020`, and `False` otherwise?

```python
#Question Number 1
df=df_spark.filter(df_spark['type']=='Movie')
df.show()
```

```python
#Question Number 2
# Your answer is completely correct!
# To strictly create a new DataFrame, we can assign it to a variable:
df_selected = df_spark.select("title", "director", "release_year")
df_selected.show()
```

```python
# Question Number 3
# How do you rename the column 'release_year' to 'year_of_release'?
# Write your code below and display the result:

New_df =df_spark.withColumnRenamed("release_year","year_of_release")
New_df.show()
```

```python
# Question Number 4
# How do you drop the 'description' column entirely from the DataFrame?
# Write your code below and display the result:

# Correction: We use .drop() to remove columns. .dropna() is for rows with nulls.
df_dropped = df_spark.drop("description")
df_dropped.show(5)
df_dropped.printSchema()
```
